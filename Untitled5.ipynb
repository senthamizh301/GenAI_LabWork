```python
import numpy as np

from datasets import load_dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    pipeline
)

import evaluate

print("Libraries imported successfully!")
```

```python
dataset = load_dataset("stanfordnlp/imdb")
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
def tokenize(example):
    return tokenizer(
        example["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

tokenized_dataset = dataset.map(tokenize, batched=True)
```

```python
train_dataset = tokenized_dataset["train"].shuffle(seed=42).select(range(500))

test_dataset = tokenized_dataset["test"].shuffle(seed=42).select(range(200))
```

```python
model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=2
)
```

```python
accuracy = evaluate.load("accuracy")

def compute_metrics(eval_pred):
    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=1)

    return accuracy.compute(
        predictions=predictions,
        references=labels
    )
```

```python
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=1,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    logging_steps=20
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics
)
```

```python
results=trainer.evaluate()
print(results)
```

```python
classifier=pipeline(
    "sentiment-analysis",
    model=trainer.model,
    tokenizer=tokenizer
)
```

```python
print(classifier("The movie was absolutely wonderful!"))
print(classifier("The movie was boring and disappointing."))
```
